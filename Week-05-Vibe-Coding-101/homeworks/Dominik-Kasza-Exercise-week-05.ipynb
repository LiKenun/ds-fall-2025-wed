```python
# Question 1: What is the median salary for Software Engineers in the United States?


df = pd.read_csv("Ask A Manager Salary Survey 2021 (Responses) - Form Responses 1.tsv", sep="\t")
mask = (
    df["Job title"].astype(str).str.lower().str.contains("software engineer", na=False)
    & df["What country do you work in?"].astype(str).str.lower().str.contains("united states|usa|us", na=False)
    & (df["Please indicate the currency"].astype(str).str.upper() == "USD")
)
salary_col = "What is your annual salary? (You'll indicate the currency in a later question. If you are part-time or hourly, please enter an annualized equivalent -- what you would earn if you worked the job 40 hours a week, 52 weeks a year.)"
df.loc[mask, salary_col] = df.loc[mask, salary_col].astype(str).str.replace(",", "", regex=False).astype(float)
print(df.loc[mask, salary_col].median())
```

```python
# Question 2: Which US state has the highest average salary for tech workers?


df = pd.read_csv("Ask A Manager Salary Survey 2021 (Responses) - Form Responses 1.tsv", sep="\t")
mask = (
    df["What industry do you work in?"].astype(str).str.contains("computing|tech|software|information|it", case=False, na=False)
    & df["What country do you work in?"].astype(str).str.lower().str.contains("united states|usa|us", na=False)
    & (df["Please indicate the currency"].astype(str).str.upper() == "USD")
)
salary_col = "What is your annual salary? (You'll indicate the currency in a later question. If you are part-time or hourly, please enter an annualized equivalent -- what you would earn if you worked the job 40 hours a week, 52 weeks a year.)"
df.loc[mask, salary_col] = df.loc[mask, salary_col].astype(str).str.replace(",", "", regex=False).astype(float)
print(df[mask].groupby("If you're in the U.S., what state do you work in?")[salary_col].mean().sort_values(ascending=False).head(1))
```

```python
# Question 3: How much does salary increase on average for each year of experience in tech?

df = pd.read_csv("Ask A Manager Salary Survey 2021 (Responses) - Form Responses 1.tsv", sep="\t")
mask = (
    df["What industry do you work in?"].astype(str).str.contains("computing|tech|software|information|it", case=False, na=False)
    & df["What country do you work in?"].astype(str).str.lower().str.contains("united states|usa|us", na=False)
    & (df["Please indicate the currency"].astype(str).str.upper() == "USD")
)
salary_col = "What is your annual salary? (You'll indicate the currency in a later question. If you are part-time or hourly, please enter an annualized equivalent -- what you would earn if you worked the job 40 hours a week, 52 weeks a year.)"
df.loc[mask, salary_col] = df.loc[mask, salary_col].astype(str).str.replace(",", "", regex=False).astype(float)
def extract_years(x):
    if pd.isna(x): return np.nan
    x = str(x)
    if "Less than" in x: return 0.5
    if "More than" in x: return 35
    n = [float(i) for i in x.replace("+","").replace("years","").replace("year","").replace(" ","").split("-") if i.isdigit()]
    return np.mean(n) if len(n)==2 else (n[0] if len(n)==1 else np.nan)
df["YearsExperience"] = df["How many years of professional work experience do you have overall?"].apply(extract_years)
x = df.loc[mask].dropna(subset=[salary_col, "YearsExperience"])
slope, intercept = np.polyfit(x["YearsExperience"], x[salary_col], 1)
print(slope)
```

```python

# Question 4: What percentage of respondents work remotely vs. in-office?


df = pd.read_csv("Ask A Manager Salary Survey 2021 (Responses) - Form Responses 1.tsv", sep="\t")

col = "What is your current work arrangement (remote vs. in-office)?"
df[col] = df[col].astype(str).str.lower().str.strip()

remote_counts = df[col].value_counts(dropna=True)
percentages = (remote_counts / remote_counts.sum()) * 100

print(percentages)
```

```python
# Question 5: Which industry (besides tech) has the highest median salary?


df = pd.read_csv("Ask A Manager Salary Survey 2021 (Responses) - Form Responses 1.tsv", sep="\t")

salary_col = "What is your annual salary? (You'll indicate the currency in a later question. If you are part-time or hourly, please enter an annualized equivalent -- what you would earn if you worked the job 40 hours a week, 52 weeks a year.)"
df[salary_col] = df[salary_col].astype(str).str.replace(",", "", regex=False).astype(float)

mask = (
    (df["Please indicate the currency"].astype(str).str.upper() == "USD")
    & (~df["What industry do you work in?"].astype(str).str.contains("computing|tech|software|information|it", case=False, na=False))
)

median_by_industry = df[mask].groupby("What industry do you work in?")[salary_col].median().sort_values(ascending=False)
print(median_by_industry.head(1))
```
